```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = df[['Total Cases', 'Active', 'Discharged', 'Deaths']].sum().reset_index()
data.columns = ['Category', 'Count']

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Plot 1: Total Cases and Discharged (larger scale)
major_cases = data[data['Category'].isin(['Total Cases', 'Discharged'])]
sns.barplot(data=major_cases, x='Category', y='Count', ax=axes[0])
axes[0].set_title('COVID-19 Total Cases and Discharged')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Number of Cases')
axes[0].ticklabel_format(style='plain', axis='y') # Prevent scientific notation

# Plot 2: Active and Deaths (smaller scale)
minor_cases = data[data['Category'].isin(['Active', 'Deaths'])]
sns.barplot(data=minor_cases, x='Category', y='Count', ax=axes[1])
axes[1].set_title('COVID-19 Active Cases and Deaths')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Number of Cases')
axes[1].ticklabel_format(style='plain', axis='y') # Prevent scientific notation

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```

```python
top10_cases = df.sort_values('Total Cases', ascending=False).head(10)

top10_cases[['State/UTs', 'Total Cases']]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10_cases,
    x='Total Cases',
    y='State/UTs'
)

plt.title('Top 10 States/UTs by Total COVID-19 Cases')
plt.xlabel('Total Cases')
plt.ylabel('State/UT')

plt.show()
```

```python
top10_deaths = df.sort_values('Deaths', ascending=False).head(10)

top10_deaths[['State/UTs', 'Deaths']]
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10_deaths,
    x='Deaths',
    y='State/UTs'
)

plt.title('Top 10 States/UTs by COVID-19 Deaths')
plt.xlabel('Number of Deaths')
plt.ylabel('State/UT')

plt.show()
```

```python
ratio_data = df[['Active Ratio', 'Discharge Ratio', 'Death Ratio']].mean().reset_index()

ratio_data.columns = ['Ratio', 'Average']

ratio_data
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=ratio_data,
    x='Ratio',
    y='Average'
)

plt.title('Average COVID-19 Ratios')
plt.xlabel('Ratio')
plt.ylabel('Average Percentage')
```

```python
top10_vaccine = df.sort_values('Total Doses', ascending=False).head(10)

top10_vaccine[['State/UTs', 'Total Doses']]
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10_vaccine,
    x='Total Doses',
    y='State/UTs'
)

plt.title('Top 10 States/UTs by Total COVID-19 Vaccination Doses')
plt.xlabel('Total Vaccination Doses')
plt.ylabel('State/UT')

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(14, 10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of COVID-19 Dataset')
plt.show()
```

```python
df['Recovery Rate'] = (df['Discharged'] / df['Total Cases']) * 100

top10_recovery = df.sort_values('Recovery Rate', ascending=False).head(10)

top10_recovery[['State/UTs', 'Total Cases', 'Discharged', 'Recovery Rate']]
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10_recovery,
    x='Recovery Rate',
    y='State/UTs'
)

plt.title('Top 10 States/UTs by COVID-19 Recovery Rate')
plt.xlabel('Recovery Rate (%)')
plt.ylabel('State/UT')

plt.show()
```
